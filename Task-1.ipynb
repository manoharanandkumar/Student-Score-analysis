```python
import pandas as pd 
import numpy as np
file_path = "C:/Users/Manoh/Downloads/student/student-mat.csv"
data = pd.read_csv(file_path, sep=";")
#Average score for G3 in math
average_score = data[ 'G3'].mean()
print(f"Average score in math (G3): {average_score}")
#Student scored above 15 in final grade G3
students_above_15 = data[data['G3'] > 15]. shape[0]
print(f"Number of students who scored above 15 in their final grade (G3): {students_above_15}")
#corelation between study time and final grade
correlation = data['studytime'].corr(data['G3'])
print(f"Correlation between Study Time and Final grade: {correlation}")
#gender has higher average final grade
avg_grade_gender = data.groupby('sex')['G3'].mean()
print(f"Average final grade (G3) by gender: {avg_grade_gender}")
```
